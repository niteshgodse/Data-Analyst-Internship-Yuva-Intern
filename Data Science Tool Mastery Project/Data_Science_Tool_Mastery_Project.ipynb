## 1. Tools Covered
Python | Pandas | NumPy | Matplotlib | Seaborn | SQL/MySQL concepts | Scikit-learn | Jupyter Notebook

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sales = pd.read_csv('tool_mastery_sales_data.csv')
sales.head()
````

## 2. Pandas — Data Inspection and Manipulation

````python
print('Shape:', sales.shape)
print(sales.info())
print('\nMissing values:')
print(sales.isnull().sum())
print('\nSummary:')
print(sales.describe())
````

````python
category_summary = sales.groupby('Category', as_index=False)['Revenue'].sum().sort_values('Revenue', ascending=False)
region_summary = sales.groupby('Region', as_index=False)['Revenue'].sum().sort_values('Revenue', ascending=False)
print(category_summary)
print(region_summary)
````

## 3. NumPy — Statistical Calculations

````python
revenue = sales['Revenue'].to_numpy()
print('Mean revenue:', np.mean(revenue))
print('Median revenue:', np.median(revenue))
print('Standard deviation:', np.std(revenue))
print('90th percentile:', np.percentile(revenue, 90))
````

## 4. Matplotlib & Seaborn — Visualization

````python
sns.set_theme(style='whitegrid')
plt.figure(figsize=(8,5))
plt.bar(category_summary['Category'], category_summary['Revenue'])
plt.title('Revenue by Category')
plt.xticks(rotation=20)
plt.ylabel('Revenue')
plt.tight_layout()
plt.show()
````

````python
plt.figure(figsize=(8,5))
sns.barplot(data=region_summary, x='Region', y='Revenue')
plt.title('Revenue by Region')
plt.tight_layout()
plt.show()

plt.figure(figsize=(8,5))
sns.histplot(data=sales, x='Revenue', kde=True)
plt.title('Revenue Distribution')
plt.tight_layout()
plt.show()

plt.figure(figsize=(8,5))
sns.scatterplot(data=sales, x='Quantity', y='Revenue', hue='Category')
plt.title('Quantity vs Revenue')
plt.tight_layout()
plt.show()
````

## 5. SQL / MySQL Practice

Example queries for the same dataset after loading it into a SQL table:

```sql
SELECT Category, SUM(Revenue) AS Total_Revenue
FROM sales
GROUP BY Category
ORDER BY Total_Revenue DESC;

SELECT Region, AVG(Revenue) AS Average_Revenue
FROM sales
GROUP BY Region;

SELECT * FROM sales
WHERE Revenue > 500
ORDER BY Revenue DESC;
```

## 6. Scikit-learn — Basic Machine Learning Exercise

````python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score

# Predict revenue using numeric order characteristics
features = ['Quantity', 'Unit_Price', 'Discount']
X = sales[features]
y = sales['Revenue']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
pred = model.predict(X_test)

print('MAE:', mean_absolute_error(y_test, pred))
print('R2:', r2_score(y_test, pred))
````

## 7. Portfolio Piece — Business Insights

````python
top_category = category_summary.iloc[0]
top_region = region_summary.iloc[0]
print(f"Top revenue category: {top_category['Category']} — {top_category['Revenue']:.2f}")
print(f"Top revenue region: {top_region['Region']} — {top_region['Revenue']:.2f}")
print(f"Total revenue: {sales['Revenue'].sum():.2f}")
````
